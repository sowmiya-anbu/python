```python
class Bus:
    def __init__(self, bus_no, route, seats):
        self.bus_no = bus_no
        self.route = route
        self.total_seats = seats
        self.available_seats = seats

    def show_bus(self):
        print("\nBus No:", self.bus_no)
        print("Route:", self.route)
        print("Available Seats:", self.available_seats)

    def book_ticket(self, seats):
        if seats <= self.available_seats:
            self.available_seats -= seats
            print(f"{seats} seat(s) booked successfully.")
        else:
            print("Seats not available.")

    def cancel_ticket(self, seats):
        self.available_seats += seats
        if self.available_seats > self.total_seats:
            self.available_seats = self.total_seats
        print(f"{seats} seat(s) cancelled successfully.")


bus = Bus(101, "Chennai - Bangalore", 40)

while True:
    print("\n===== BUS BOOKING SYSTEM =====")
    print("1. Bus Details")
    print("2. Book Ticket")
    print("3. Cancel Ticket")
    print("4. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        bus.show_bus()

    elif choice == 2:
        seats = int(input("Enter Number of Seats: "))
        bus.book_ticket(seats)

    elif choice == 3:
        seats = int(input("Enter Number of Seats to Cancel: "))
        bus.cancel_ticket(seats)

    elif choice == 4:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```
