```python
class Library:
    def __init__(self, book_list):
        self.books = book_list

    def display(self):
        print("\nAvailable Books:")
        for i in self.books:
            print(i)

    def borrow(self):
        book_name = input("Enter the book name: ")
        if book_name in self.books:
            self.books.remove(book_name)
            print("Enjoy your reading!")
        else:
            print("Currently not available.")

    def return_book(self):
        book_name = input("Enter the book name: ")
        if book_name in self.books:
            print("Book already exists in library.")
        else:
            self.books.append(book_name)
            print("Thank you! Come again.")

    def add_book(self):
        book_name = input("Enter new book name: ")
        if book_name in self.books:
            print("Book already exists.")
        else:
            self.books.append(book_name)
            print("Book added successfully.")

    def search_book(self):
        book_name = input("Enter book name to search: ")
        if book_name in self.books:
            print(book_name, "is available in the library.")
        else:
            print(book_name, "is not available.")
```

```python
my_books = [
    "python", "c", "c++", "java", "html", "css",
    "javascript", "sql", "mysql", "php", "django",
    "flask", "react", "data science", "machine learning",
    "artificial intelligence", "dbms", "operating system",
    "computer networks", "cyber security"
]

obj = Library(my_books)
```

```python
while True:
    print("\n===== Library Management System =====")
    print("1. Display Books")
    print("2. Borrow Book")
    print("3. Return Book")
    print("4. Add Book")
    print("5. Search Book")
    print("6. Exit")

    user = int(input("Enter your choice: "))

    if user == 1:
        obj.display()

    elif user == 2:
        obj.borrow()

    elif user == 3:
        obj.return_book()

    elif user == 4:
        obj.add_book()

    elif user == 5:
        obj.search_book()

    elif user == 6:
        print("Thank you for using Library Management System.")
        break

    else:
        print("Invalid Choice! Please try again.")
```
