```python
# Hybrid Inheritance
class parent:
    def f1(self):
        print("Function1 from parent class")
    def f2(self):
        print("Function2 from parent class")
class child1(parent):
    def f3(self):
        print("Function3 from child1 class")
    def f4(self):
        print("Function4 from child1 class")
class child2(child1):
    def f5(self):
        print("Function5 from child2 class")
    def f6(self):
        print("Function6 from child2 class")
class child3(child1):
    def f7(self):
        print("Function7 from child3 class")
    def f8(self):
        print("Function8 from child3 class")
```

```python
obj = child3()
```

```python
obj.f1()
```

```python
obj.f3()
```

```python
obj.f7()
```

```python
obj.f2()
```

```python
obj.f4()
```

```python
obj.f8()
```

```python
class parent:
    def f1(self):
        self.price = 5000
        print("price value is",self.price)
class child(parent):
    def f1(self):
        self.price = 10000
        print("price value is",self.price)
        
```

```python
obj = child()
```

```python
obj.f1()
```

```python
new_obj = parent()
```

```python
new_obj.f1()
```
